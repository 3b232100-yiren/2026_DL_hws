```python
import numpy as np


def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):

    x = x - np.max(x, axis=-1, keepdims=True)
    return np.exp(x) / np.sum(np.exp(x), axis=-1, keepdims=True)

def cross_entropy_error(y_hat, t):

    if y_hat.ndim == 1:
        t = t.reshape(1, t.size)
        y_hat = y_hat.reshape(1, y_hat.size)
    batch_size = y_hat.shape[0]
    return -np.sum(t * np.log(y_hat + 1e-7)) / batch_size
```

# 自行產生五筆輸入資料 X 及真實結果 Y

```python
X = np.array([
    [1.0, -2.0, 0.5, 1.5, -0.5, 2.0],
    [-0.5, 1.0, 1.2, -1.1, 0.3, 0.4],
    [0.2, 0.3, -0.4, 1.1, -0.8, 0.9],
    [-1.2, -0.5, 0.8, 0.2, 1.0, -0.3],
    [0.5, 0.5, 0.5, -0.5, -0.5, -0.5]
])

Y = np.array([
    [0, 0, 1],
    [0, 1, 0],
    [1, 0, 0],
    [0, 0, 1],
    [0, 1, 0]
])

print("=== 輸入資料與真實結果形狀 ===")
print("X 形狀:", X.shape)
print("Y 形狀:", Y.shape)
```

# 設定各層權重與偏差

```python
W1 = np.array([
    [0.1, -0.2, 0.3, -0.1, 0.2],
    [0.4, 0.1, -0.3, 0.2, -0.4],
    [-0.1, 0.3, 0.2, 0.1, -0.2],
    [0.2, -0.3, 0.1, -0.2, 0.3],
    [-0.3, 0.2, -0.1, 0.3, 0.1],
    [0.1, 0.2, -0.2, -0.1, 0.4]
])
b1 = np.array([0.1, -0.1, 0.2, 0.0, 0.1])

W2 = np.array([
    [0.2, -0.1, 0.3, 0.1],
    [-0.3, 0.2, 0.1, -0.2],
    [0.1, 0.4, -0.2, 0.3],
    [-0.2, -0.3, 0.2, 0.1],
    [0.4, 0.1, -0.1, -0.3]
])
b2 = np.array([0.0, 0.2, -0.1, 0.1])

W3 = np.array([
    [0.1, 0.3, -0.2, 0.2],
    [-0.2, 0.1, 0.4, -0.1],
    [0.3, -0.3, 0.1, 0.2],
    [-0.1, 0.2, -0.3, 0.4]
])
b3 = np.array([0.1, 0.0, -0.2, 0.1])

W4 = np.array([
    [0.4, -0.2, 0.1],
    [-0.1, 0.3, 0.2],
    [0.2, 0.1, -0.3],
    [-0.3, 0.2, 0.4]
])
b4 = np.array([0.1, -0.1, 0.2])


print("=== 各層權重矩陣、偏差向量與計算結果形狀 ===\n")
# 第一隱藏層
z1 = np.dot(X, W1) + b1
h1 = relu(z1)
print("【第一隱藏層】")
print("W1 陣列:\n", W1, f"| 形狀: {W1.shape}")
print("b1 陣列:", b1, f"| 形狀: {b1.shape}")
print("調整後加權和 z1 形狀:", z1.shape)
print("輸出 h1 形狀:", h1.shape, "\n")

# 第二隱藏層
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)
print("【第二隱藏層】")
print("W2 陣列:\n", W2, f"| 形狀: {W2.shape}")
print("b2 陣列:", b2, f"| 形狀: {b2.shape}")
print("調整後加權和 z2 形狀:", z2.shape)
print("輸出 h2 形狀:", h2.shape, "\n")

# 第三隱藏層
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)
print("【第三隱藏層】")
print("W3 陣列:\n", W3, f"| 形狀: {W3.shape}")
print("b3 陣列:", b3, f"| 形狀: {b3.shape}")
print("調整後加權和 z3 形狀:", z3.shape)
print("輸出 h3 形狀:", h3.shape, "\n")

# 輸出層
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)
print("【輸出層】")
print("W4 陣列:\n", W4, f"| 形狀: {W4.shape}")
print("b4 陣列:", b4, f"| 形狀: {b4.shape}")
print("調整後加權和 z4 形狀:", z4.shape)
print("輸出 y_hat 形狀:", y_hat.shape, "\n")

# 5. 計算整批平均交叉熵誤差
batch_loss = cross_entropy_error(y_hat, Y)

print("=== 批次學習最終誤差 ===")
print("預測結果 y_hat:\n", y_hat)
print("\n真實結果 Y:\n", Y)
print(f"\nCross Entropy Error (整批平均交叉熵誤差): {batch_loss:.4f}")
```
