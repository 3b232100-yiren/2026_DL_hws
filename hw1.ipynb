```python
import numpy as np

# 1. 定義激發函數
def relu(x):
    return np.maximum(0, x)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def softmax(x):
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x)

# 2. 定義 Cross Entropy Error
def cross_entropy_error(y_hat, y):
    delta = 1e-7
    return -np.sum(y * np.log(y_hat + delta))

# 輸入資料
X = np.array([1.0, -2.0, 0.5, 1.5, -0.5, 2.0])
print("輸入資料 X:", X)
print("輸入層形狀 (Shape):", X.shape)
print("特徵值個數:", X.size, "\n")

# --- 第一隱藏層 ---
W1 = np.array([
    [0.1, -0.2, 0.3, -0.1, 0.2],
    [0.4, 0.1, -0.3, 0.2, -0.4],
    [-0.1, 0.3, 0.2, 0.1, -0.2],
    [0.2, -0.3, 0.1, -0.2, 0.3],
    [-0.3, 0.2, -0.1, 0.3, 0.1],
    [0.1, 0.2, -0.2, -0.1, 0.4]
])
b1 = np.array([0.1, -0.1, 0.2, 0.0, 0.1])
z1 = np.dot(X, W1) + b1
h1 = relu(z1)

print("【第一隱藏層】")
print("W1 形狀:", W1.shape, "| 參數個數:", W1.size)
print("b1 形狀:", b1.shape, "| 參數個數:", b1.size)
print("調整後加權和 z1:", z1, "| 形狀:", z1.shape)
print("輸出向量 h1 (ReLU):", h1, "| 形狀:", h1.shape, "\n")

# --- 第二隱藏層 ---
W2 = np.array([
    [0.2, -0.1, 0.3, 0.1],
    [-0.3, 0.2, 0.1, -0.2],
    [0.1, 0.4, -0.2, 0.3],
    [-0.2, -0.3, 0.2, 0.1],
    [0.4, 0.1, -0.1, -0.3]
])
b2 = np.array([0.0, 0.2, -0.1, 0.1])
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

print("【第二隱藏層】")
print("W2 形狀:", W2.shape, "| 參數個數:", W2.size)
print("b2 形狀:", b2.shape, "| 參數個數:", b2.size)
print("調整後加權和 z2:", z2, "| 形狀:", z2.shape)
print("輸出向量 h2 (ReLU):", h2, "| 形狀:", h2.shape, "\n")

# --- 第三隱藏層 ---
W3 = np.array([
    [0.1, 0.3, -0.2, 0.2],
    [-0.2, 0.1, 0.4, -0.1],
    [0.3, -0.3, 0.1, 0.2],
    [-0.1, 0.2, -0.3, 0.4]
])
b3 = np.array([0.1, 0.0, -0.2, 0.1])
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

print("【第三隱藏層】")
print("W3 形狀:", W3.shape, "| 參數個數:", W3.size)
print("b3 形狀:", b3.shape, "| 參數個數:", b3.size)
print("調整後加權和 z3:", z3, "| 形狀:", z3.shape)
print("輸出向量 h3 (Sigmoid):", h3, "| 形狀:", h3.shape, "\n")

# --- 輸出層 ---
W4 = np.array([
    [0.4, -0.2, 0.1],
    [-0.1, 0.3, 0.2],
    [0.2, 0.1, -0.3],
    [-0.3, 0.2, 0.4]
])
b4 = np.array([0.1, -0.1, 0.2])
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)

print("【輸出層】")
print("W4 形狀:", W4.shape, "| 參數個數:", W4.size)
print("b4 形狀:", b4.shape, "| 參數個數:", b4.size)
print("調整後加權和 z4:", z4, "| 形狀:", z4.shape)
print("輸出結果 y_hat (Softmax):", y_hat, "| 形狀:", y_hat.shape, "\n")

# --- 誤差計算 ---
y = np.array([0, 0, 1])
loss = cross_entropy_error(y_hat, y)

print("====================================")
print("真實結果 y:", y)
print("預測結果 y_hat:", y_hat)
print("Cross Entropy Error:", loss)
print("====================================")
```

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    c = np.max(x)
    exp_x = np.exp(x - c)
    return exp_x / np.sum(exp_x)

W1 = np.array([
    [0.1, -0.2, 0.3, -0.1, 0.2],
    [0.4, 0.1, -0.3, 0.2, -0.4],
    [-0.1, 0.3, 0.2, 0.1, -0.2],
    [0.2, -0.3, 0.1, -0.2, 0.3],
    [-0.3, 0.2, -0.1, 0.3, 0.1],
    [0.1, 0.2, -0.2, -0.1, 0.4]
])
b1 = np.array([0.1, -0.1, 0.2, 0.0, 0.1])

W2 = np.array([
    [0.2, -0.1, 0.3, 0.1],
    [-0.3, 0.2, 0.1, -0.2],
    [0.1, 0.4, -0.2, 0.3],
    [-0.2, -0.3, 0.2, 0.1],
    [0.4, 0.1, -0.1, -0.3]
])
b2 = np.array([0.0, 0.2, -0.1, 0.1])

W3 = np.array([
    [0.1, 0.3, -0.2, 0.2],
    [-0.2, 0.1, 0.4, -0.1],
    [0.3, -0.3, 0.1, 0.2],
    [-0.1, 0.2, -0.3, 0.4]
])
b3 = np.array([0.1, 0.0, -0.2, 0.1])

W4 = np.array([
    [0.4, -0.2, 0.1],
    [-0.1, 0.3, 0.2],
    [0.2, 0.1, -0.3],
    [-0.3, 0.2, 0.4]
])
b4 = np.array([0.1, -0.1, 0.2])

def dnn(x1, x2, x3, x4, x5, x6):
    x = np.array([x1, x2, x3, x4, x5, x6])

    z1 = np.dot(x, W1) + b1
    h1 = relu(z1)

    z2 = np.dot(h1, W2) + b2
    h2 = relu(z2)

    z3 = np.dot(h2, W3) + b3
    h3 = sigmoid(z3)

    z4 = np.dot(h3, W4) + b4
    y_hat = softmax(z4)

    max_class = int(np.argmax(y_hat)) + 1

    return f"預測機率 y_hat: {y_hat}\n👉 機率最高的類別為：第 {max_class} 類"

!pip install -q gradio
import gradio as gr

inputs_list = [
    gr.Number(value=1.0, label="輸入特徵值 x1"),
    gr.Number(value=-2.0, label="輸入特徵值 x2"),
    gr.Number(value=0.5, label="輸入特徵值 x3"),
    gr.Number(value=1.5, label="輸入特徵值 x4"),
    gr.Number(value=-0.5, label="輸入特徵值 x5"),
    gr.Number(value=2.0, label="輸入特徵值 x6")
]

iface = gr.Interface(
    fn=dnn,
    inputs=inputs_list,
    outputs="text",
    title="DNN 多隱藏層神經網路分類器",
    description="請輸入 6 個特徵值 (x1 ~ x6)，系統將會透過 DNN 進行運算，並輸出機率最高的類別 (1, 2 或 3)！"
)

iface.launch()
```
